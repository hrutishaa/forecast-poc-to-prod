# 01 · PoC: can we beat a seasonal-naive forecast?
This is the *before* picture: a quick exploratory notebook. Everything that worked here was moved into the `forecast/` package, where it is tested, configured by `config.toml`, versioned, and served by FastAPI.

```python
import sys; sys.path.insert(0, '..')
from forecast.data import load_series
from forecast.models import SeasonalNaive, GBMForecaster
from forecast.backtest import backtest

y = load_series('retail_sales')
y.plot(title='retail_sales');
```

```python
base = backtest(y, lambda: SeasonalNaive(12), horizon=12, folds=3)
model = backtest(y, lambda: GBMForecaster(), horizon=12, folds=3)
print(f"baseline MAPE {base['mape']:.2f}%  |  GBM MAPE {model['mape']:.2f}%")
```
